**Ventaja clave:** No necesitas mallar manualmente el dominio.

## 📚 ¿Qué son las Ecuaciones de Navier-Stokes?

Son las ecuaciones que describen cómo fluyen los líquidos y gases:

$$
\rho\left(\frac{\partial \mathbf{v}}{\partial t} + \mathbf{v}\cdot\nabla\mathbf{v}\right) = -\nabla p + \mu\nabla^2\mathbf{v} + \mathbf{f}
$$

**Historia:**
- **1822:** Claude-Louis Navier (ingeniero francés) las deriva incluyendo viscosidad
- **1845:** George Gabriel Stokes (matemático irlandés) las re-deriva mejor

**Problema del Milenio:**
- Nadie ha probado si siempre tienen solución suave en 3D
- Premio: $1,000,000 USD (Clay Mathematics Institute)
- Estado: Sin resolver después de 200 años

## 🧠 ¿Qué es un PINN?

**Physics-Informed Neural Network** = Red neuronal que aprende física.

En lugar de solo ajustar datos, **respeta las ecuaciones diferenciales**.

### Función de Pérdida

```python
loss = loss_datos + λ * loss_ecuacion + λ_bc * loss_frontera
```

Donde:
- `loss_datos`: Error en mediciones (si existen)
- `loss_ecuacion`: Violación de Navier-Stokes
- `loss_frontera`: Violación de condiciones de frontera
